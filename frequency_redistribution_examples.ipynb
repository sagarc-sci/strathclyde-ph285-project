# Redistribution of Gamma wavelengths

During Hydrogen burning of nuclear fusion in core of a star energy is emitted in the form of Gamma-ray photons.
As these photons move through the stellar atmosphere they are wavelength shifted as a result of number of processes
such as Compton scattering on inelastic collisions with electrons, Doppler shift on collisions with fast moving electrons,
losing energy while climbing out of Gravitational potential well, etc. In this simulation, we use a $5.49~MeV$ Gamma photon
emitted during PP-I chain when Deuterium nucleus captures a proton.

Of these processes, Compton scattering can shift the wavelength by up to $4.85~pm$ ($= \frac{2 h}{m_e c}$)
where as Doppler shift contributes to $~1.30 \%$ of wavelength at the simulated gas temperature of $1M~K$.
Additionally, free-free absorption by electrons in potential well of nearby protons converts nuclear photons
into thermal photons during re-emission following Planck's law for Black Body radiation. However, at Gamma wavelengths
cross section of this process has no significant impact.

The following simulation demsonstrates shift in Gamma wavelengths to soft X-ray wavelengths as they
pass through a $5~m$ thick slab of mostly ionised Hydrogen gas at $1M~K$ with number density of $1e30~m^{-3}$.
Note that environment in stellar core is much more extreme than being simulated here (e.g. higher electron density due to
nuclear electrons) and other processes come into play. However, the demonstration is useful to understand assumptions made
in spectral synthesis code.

The mean number of steps to escape is roughly what we expect it to be for photons that can only escape at the right
boundary of the atmosphere ($\approx \frac{\tau^{2}}{2} = \frac{(R \alpha)^{2}}{2}$).

# Redistribution of UV wavelengths

In this demonstration we show the increased impact of Doppler shifting and free-free re-emissions at UV wavelengths.
Following the simulation at X-ray wavelengths, Compton scattering has an insignificant impact at this wavelength range.
However, at a temperature of $1M~K$, electrons have a sigificant thermal energy as a fraction of UV photon's energy.
Collisions with these high energy electrons can Doppler shift the wavelengths in either direction.
We also observe an increase in free-free interactions as the cross section of this process increases proportional to $\lambda^{3}$.
However, the shifting in wavelength is largely due to Doppler kicks from high energy electrons.

As we move into higher wavelengths, free-free re-emission becomes dominant and spectrum of redistributed wavelengths begins
to resemble Black Body radiation spectrum as more and more nuclear photons are converted into thermal photons.
This forms the basis for our spectral synthesis code which treats the photons entering the simulated layer of atmosphere
as thermal photons from Black Body source.
