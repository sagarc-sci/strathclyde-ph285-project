# Spectral synthesis in spherical atmosphere

In this Stopped Random Walk Monte-Carlo simulation we synthesise the spectrum of $110$ grid units (say $110~m$) thick pure Hydrogen spherical atmosphere with a number density of $10^{22}~m^{-3}$ and a temperature of $20000~K$ at the inner edge exponentially dropping to $10^{20} ~m^{-3}$ and $10000~K$ at the outer edge.

Following the results from frequency redistribution experiments, we treat the photons as emerging from a Black Body radiation source at same temperature as inner edge of the atmosphere.

The resulting spectrum shows Hydrogen absorption lines at expected wavelengths and shows a significant drop in wavelengths below the Hydrogen ionisation wavelength. We also see a drop in overall intensity at higher angles as photons get absorbed more often in colder atmosphere demonstrating the _Limb Darkening_ effect.
